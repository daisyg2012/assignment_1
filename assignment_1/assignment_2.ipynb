```python
import numpy as np
"""
This function calculates f(x) = x^3 - x^2 - 1

Parameters: x, integer value

Returns:
f(x)
"""
def f(x):
    return x**3-x**2-1

"""
This function calculates f'(x) = 3x^2 - 2x

Parameters: x, integer value

Returns:
f'(x)
"""
def df(x):
    return 3*x**2 - 2*x
"""Part 2:

This function performs a Newton iteration

Parameters: f, df, x0, epsilon, max_iter

Returns:
value of Xi or None"""
def newton(f, df, x0, epsilon=1e-6, max_iter=30):
    i = 0
    x = x0
    while i < max_iter:
        if(np.abs(f(x)) < epsilon):
            print("Found root in " + str(i) + " iterations")
            return x
        
        x = x - f(x)/df(x)
        
        i+=1
    print("Iteration failed")
    return None
"""Part 3: different values of X0 + different epsilon, increasing iterations by 1"""
newton(f, df, 2)
newton(f, df, 1)

newton(f, df, 1, epsilon=1e-8)


        
```
